# QM9 combined sampling plots

Samplers (`samples-full/samples-old`) + branching flows (`qm9_permutation_best` / `qm9_no_permutation_best`) on one **num_steps** axis. One PDF per metric under `plots/qm9_sampling/` — edit paths / checkpoint in the next cell.

```python
from __future__ import annotations

import sys
from pathlib import Path

# Repo root + fresh import of qm9_sampling_plot_utils
_here = Path.cwd().resolve()
_root = _here
if not (_root / "qm9_sampling_plot_utils.py").is_file():
    for _p in _here.parents:
        if (_p / "qm9_sampling_plot_utils.py").is_file():
            _root = _p
            break
if str(_root) not in sys.path:
    sys.path.insert(0, str(_root))
if "qm9_sampling_plot_utils" in sys.modules:
    del sys.modules["qm9_sampling_plot_utils"]

from qm9_sampling_plot_utils import (
    collect_combined_num_steps_grid,
    configure_matplotlib,
    plot_all_metrics,
)

configure_matplotlib()

ROOT = _root
NUM_STEPS_ROOT = ROOT / "samples-full" / "samples-old"
SAMPLES_OLD_CKPT_K = 1350
BF_GRID = ROOT / "bf_grid_search_large" / "qm9_sampling_grid"
OUT_DIR = ROOT / "plots" / "qm9_sampling"
# "valid_only" or "all_samples" (subdir under each n_gen_* run; not "all")
MODE = "all_samples"

df = collect_combined_num_steps_grid(
    NUM_STEPS_ROOT,
    BF_GRID / "qm9_permutation_best",
    BF_GRID / "qm9_no_permutation_best",
    mode=MODE,
    ablation_layout="samples_old",
    samples_old_ckpt_k=SAMPLES_OLD_CKPT_K,
)

OUT_DIR.mkdir(parents=True, exist_ok=True)
plot_all_metrics(
    df,
    x_col="num_steps",
    x_label="Steps",
    file_prefix="combined_num_steps_grid",
    out_dir=OUT_DIR,
    layout="compact",
)
print(f"Wrote combined plots under {OUT_DIR.resolve()}")
```
